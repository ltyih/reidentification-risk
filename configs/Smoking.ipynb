```python
import random 
import numpy as np
import pandas as pd
import names
```

```python
def smoking(n):
    p = 0.11
    smoke_yn = random.choices(["yes","no"], [p, 1-p], k=n)
    return(smoke_yn)
```

```python
def sample_province(n):
    provinces = ["ON","QB","BC", "AB" , "MB", "SK", "NS", "NB", "NL", "PEI", "NWT", "YK", "NV"]
    weights = [0.3845, 0.2298, 0.1352, 0.1152, 0.0363, 0.0306,0.0262, 0.0209, 0.0138, 0.0042, 0.0011, 0.0011, 0.0010]
    results = random.choices(provinces, weights=weights, k=n)
    return(results)
```

```python
def gender(n):
    p = 0.05
    gend = random.choices(["male","female","other"], [(1-p)/2, (1-p)/2, p], k=n)
    return(gend)
```

```python
def cancer(n):
    p = 0.05
    gend = random.choices(["yes","no",], [p, 1-p], k=n)
    return(gend)
```

```python
def depression_score(n):
    dep = np.random.poisson(lam=3, size=n)
    return(dep*10)
```

```python
def dataset_1(n):
    smoke = smoking(n)
    gend = gender(n)
    last_names = [None]*n
    for i in range(n-1):
        last_names[i] =names.get_last_name()
    prov = sample_province(n)
    data1 = np.column_stack((last_names,prov,gend, smoke))
    return(data1)
```

```python
def dataset_2(data1,n,k):
    #data_1 = an array generated by dataset_1
    # n = number of entries in dataset_2 
    # k = number of overlapping entries between dataset 1 and dataset 2 
    # k < n and k > length(data1)
    if k > n:
        return("k must be smaller than or equal to n")
    elif k > len(data1):
        return("k must be smaller than or equal to the number of entries in data1")
    else:
        # Randomly select the overlapping data between the two datasets
        rng = np.random.default_rng()
        ints = random.sample(range(0, len(data1)), k)
        mat1 = data1[ints]
        # Transform the existing data into the correct form for dataset 2
        # Remove name column and province column
        mat1 = mat1[:, 2:len(data1)-1]
        # Add cancer data and depression data
        canc1 = ["yes"]*k
        dep1 = depression_score(k)
        mat2 = np.column_stack((mat1, canc1, dep1))
        # Generate new data for dataset 2
        gend = gender(n-k)
        smoke = smoking(n-k)
        canc2= cancer(n-k)
        dep2 = depression_score(n-k)
        mat3 = np.column_stack((gend, smoke, canc2, dep2))
        # Combine existing data with new data and scramble the rows 
        result1 = np.vstack([mat2, mat3])
        rng = np.random.default_rng()
        result2 = rng.permutation(result1)
        return(result2)
```

```python
def create_data(n1,n2,k):
    #n1 = number of individuals in dataset 1 
    #n2 = number of individuals in dataset 2
    #k = number of overlapping individuals
    if k > n1:
        return("k must be smaller than or equal to n1")
    elif k > n2:
        return("k must be smaller than or equal to n2")
    else:  
        data1 = dataset_1(n1)
        data2 = dataset_2(data1, n2,k)
        df1 = pd.DataFrame(data1, columns=['Last Name','Province','Gender', 'Smokes'])
        df2 = pd.DataFrame(data2, columns = ['Gender', 'Smokes', 'Has cancer', 'Depression Score'])
        return(df1,df2)
```

```python
data1, data2 = create_data(40,30,10)
```
